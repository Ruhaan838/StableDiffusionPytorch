```python
from Models import model_loader as ml
from Pipe import pipeline

import torch
from PIL import Image
from transformers import CLIPTokenizer

DEVICE = "cpu"

ALLOW_CUDA = False
ALLOW_MPS = False

if torch.cuda.is_available() and ALLOW_CUDA:
    DEVICE = "cuda"

print(f"Using device: {DEVICE}")

tokenizer = CLIPTokenizer("./data/vocab.json", merges_file="./data/merges.txt") 
model_file = "./data/v1-5-pruned-emaonly.ckpt" 
models = ml.preload_models_from_standard_weights(model_file, DEVICE) 
```

```python
prompt = "astronomer on moon with one person without space shut, high quality photo, ultra sharp, 8k quality"  
uncond_prompt = ""   
do_cfg = True   
cfg_scale = 8  

strength = 0.9  


sampler = "ddpm"  
num_inference_steps = 50  
seed = 42  

output_image = pipeline.generate(
    prompt=prompt,
    uncond_prompt=uncond_prompt,
    strength=strength,
    do_cfg=do_cfg,
    cfg_scale=cfg_scale,
    n_inference_steps=num_inference_steps,
    seed=seed,
    models=models,
    device=DEVICE,
    idle_device="cpu",
    tokenizer=tokenizer,
)

Image.fromarray(output_image)
```
